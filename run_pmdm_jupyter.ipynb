# PMDM run interactively on Jupyter

The examples below are based on the PMDM github page:  
https://github.com/Layne-Huang  
and should work if the user is in the same directory as data, logs and configs.  These folders are created by the build_pmdm_container.sh script

Inference of all pockets:  
https://github.com/Layne-Huang/PMDM#sample-molecules-for-all-pockets-in-the-test-set

```python
@run sample_batch.py --ckpt data/crossdocked_pocket10/500.pt --num_samples 10 --sampling_type generalized 
```

Inference for a given pocket:  
https://github.com/Layne-Huang/PMDM#sample-molecules-for-given-customized-pockets

```python
%run sample_frag.py --ckpt data/crossdocked_pocket10/500.pt --pdb_path ./data/crossdocked_pocket10/PADI4_HUMAN_1_663_0/4x8c_A_rec_4x8g_3z0_lig_tt_min_0_pocket10.pdb --mol_file ./data/crossdocked_pocket10/PADI4_HUMAN_1_663_0/4x8c_A_rec_4x8g_3z0_lig_tt_min_0.sdf --keep_index 4 5 10 11 12 13 14 --num_atom 18 --num_samples 20 --sampling_type generalized 
```
